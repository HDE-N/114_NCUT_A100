```python
import os
import pandas as pd
from scipy.ndimage import median_filter

def process_sensor_data(input_dir='./test_data/', output_dir='./test_data_2/', kernel_size=5):
    """
    讀取 input_dir 中的 CSV 檔案，對加速度與陀螺儀數據進行中值濾波，
    並將結果存入 output_dir。
    """
    # 確保輸出目錄存在
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        print(f"已建立輸出目錄: {output_dir}")

    # 定義需要濾波的關鍵字 (加速度與陀螺儀)
    target_keywords = ['acceleration', 'gyroscope']
    
    # 取得資料夾內所有 CSV 檔案
    files = [f for f in os.listdir(input_dir) if f.endswith('.csv')]
    
    if not files:
        print(f"在 {input_dir} 中找不到任何 CSV 檔案。")
        return

    print(f"開始處理，共發現 {len(files)} 個檔案...")

    for filename in files:
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)
        
        try:
            # 讀取 CSV
            df = pd.read_csv(input_path)
            
            # 找出所有包含 'acceleration' 或 'gyroscope' 的欄位
            # 例如: acceleration_X, gyroscope_Z 等
            filter_cols = [col for col in df.columns if any(k in col for k in target_keywords)]
            
            if filter_cols:
                # 對每個目標欄位應用中值濾波
                # mode='nearest' 可避免邊界處因補零而產生的誤差
                for col in filter_cols:
                    df[col] = median_filter(df[col], size=kernel_size, mode='nearest')
                
                # 儲存結果
                df.to_csv(output_path, index=False)
                print(f"[成功] 已儲存: {filename}")
            else:
                print(f"[略過] {filename} (未找到相關欄位)")
                
        except Exception as e:
            print(f"[錯誤] 處理 {filename} 時發生錯誤: {e}")

    print("\n所有作業已完成！")

# 執行主程式
if __name__ == "__main__":
    # 您可以在此調整濾波窗口大小 (kernel_size)，預設為 5
    process_sensor_data()
```
