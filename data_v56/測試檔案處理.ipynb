```python
# 將['./test_data_old', './test_data_other']內的所有檔案複製到test_data下
# 供後續測試程式調用

import shutil
import os

def copy_contents_to_target():
    # 定義來源資料夾與目標資料夾
    source_dirs = ['./test_data_old', './test_data_other']
    target_dir = './test_data'

    # 確保目標目錄存在
    if not os.path.exists(target_dir):
        os.makedirs(target_dir)
        print(f"已建立目標目錄：{target_dir}")

    for src in source_dirs:
        if os.path.isdir(src):
            print(f"正在處理資料夾: {src}")
            # 遍歷來源資料夾中的所有檔案與子資料夾
            for item in os.listdir(src):
                s_path = os.path.join(src, item)
                t_path = os.path.join(target_dir, item)

                try:
                    if os.path.isdir(s_path):
                        # 如果子項目是資料夾，複製整個目錄
                        # 若目標已存在同名目錄，此處會報錯，可根據需求調整
                        shutil.copytree(s_path, t_path, dirs_exist_ok=True)
                    else:
                        # 如果子項目是檔案，直接複製
                        shutil.copy2(s_path, t_path)
                    print(f"  已複製: {item}")
                except Exception as e:
                    print(f"  複製 {item} 時出錯: {e}")
        else:
            print(f"錯誤: 找不到來源資料夾 {src} 或它不是一個目錄。")

if __name__ == "__main__":
    copy_contents_to_target()
```
