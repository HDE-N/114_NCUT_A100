```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns


def plot_enlarged_confusion_matrix():
    # 根據圖片輸入混淆矩陣數據
    # 順序為：notTired (0), Tired (1), Other (2)
    # cm = np.array([[463, 7, 1], [22, 259, 5], [6, 0, 751]])
    # cm = np.array([[17, 42, 0], [0, 7, 0], [0, 0, 66]])
    # cm = np.array([[201, 130, 10], [5, 10, 0], [0, 0, 328]])
    # cm = np.array([[53, 6, 0], [2, 5, 0], [0, 0, 66]])
    # cm = np.array([[252, 77, 12], [7, 8, 0], [0, 0, 328]])
    cm = np.array([[405, 64, 2], [35, 247, 4], [0, 0, 757]])
    
    labels = ["notTired (0)", "Tired (1)", "Other (2)"]

    # 設定畫布大小
    plt.figure(figsize=(24, 16))

    # 設定字體大小參數
    FONT_SIZE_LABELS = 50  # 座標軸名稱字體
    FONT_SIZE_TICKS = 50  # 座標軸刻度字體
    FONT_SIZE_ANNOT = 70  # 矩陣內數字字體

    # 繪製熱圖
    # fmt='d' 表示以整數顯示，annot_kws 用於調整矩陣內文字大小
    ax = sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=labels,
        yticklabels=labels,
        annot_kws={"size": FONT_SIZE_ANNOT},
        cbar=True,
    )

    # 調整顏色條（Colorbar）的刻度字體大小
    cbar = ax.collections[0].colorbar
    cbar.ax.tick_params(labelsize=FONT_SIZE_TICKS)

    # 設定座標軸標籤與字體大小
    plt.xlabel("Predicted Label", fontsize=FONT_SIZE_LABELS, fontweight="bold")
    plt.ylabel("True Label", fontsize=FONT_SIZE_LABELS, fontweight="bold")

    # 設定座標軸刻度字體大小
    plt.xticks(fontsize=FONT_SIZE_TICKS)
    plt.yticks(fontsize=FONT_SIZE_TICKS, rotation=0)  # rotation=0 讓 Y 軸標籤保持水平

    # 調整佈局避免文字被切到
    plt.tight_layout()

    plt.savefig("confusion_matrix_enlarged.png", dpi=300, bbox_inches="tight")

    # 顯示圖表
    # plt.show()

if __name__ == "__main__":
    plot_enlarged_confusion_matrix()
```
